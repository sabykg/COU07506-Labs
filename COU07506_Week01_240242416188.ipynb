Task 1.1 – My First Notebook
Generating and Plotting Elementary Sequences

```python
n = np.arange(-5, 11)

delta = (n == 0).astype(float)
step = (n >= 0).astype(float)
ramp = n * step
expo = 0.8**n * step

fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(axes, [delta, step, ramp, expo],
                        ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

Task 1.2 – Changing the Parameters

Part A: Exponential Signals

```python
n = np.arange(-5, 11)
step = (n >= 0).astype(float)

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(13, 3))

for ax, a in zip(axes, a_values):
    x = (a**n) * step
    ax.stem(n, x)
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.tight_layout()
plt.show()
```

Part B: Cosine Signals

Comparing the periodicity of two cosine signals.

```python
n = np.arange(0, 41)

x1 = np.cos(2 * np.pi * n / 8)
x2 = np.cos(2 * np.pi * n / 7.5)

fig, axes = plt.subplots(2, 1, figsize=(10, 6))

axes[0].stem(n, x1)
axes[0].set_title("Cosine Signal: N = 8")
axes[0].set_xlabel("n")
axes[0].set_ylabel("Amplitude")
axes[0].grid(True)

axes[1].stem(n, x2)
axes[1].set_title("Cosine Signal: N = 7.5")
axes[1].set_xlabel("n")
axes[1].set_ylabel("Amplitude")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

**Analysis:**

Both cosine signals repeat exactly because both are periodic discrete-time signals.

- The first signal, cos(2πn/8), has a fundamental period of 8 samples.
- The second signal, cos(2πn/7.5), has a fundamental period of 15 samples.

Therefore, both signals repeat exactly, but they have different fundamental periods.
